```python
import pandas as pd
from sklearn.metrics import cohen_kappa_score

# 1. Load both datasets
dora = pd.read_csv("dora_gold_annotations_completed.csv")
janice = pd.read_csv("janice_shared_annotations.csv")

# 2. Merge the datasets based on headline_id (automatically find the 100 shared headlines)
merged = dora.merge(janice, on="headline_id", suffixes=("_dora", "_janice"))

print(f"Successfully merged {len(merged)} shared headlines.\n")

# 3. Calculate Cohen's Kappa for Aspect
aspect_kappa = cohen_kappa_score(merged["aspect_dora"], merged["aspect_janice"])

# 4. Calculate Cohen's Kappa for Sentiment
sentiment_kappa = cohen_kappa_score(merged["sentiment_dora"], merged["sentiment_janice"])

# 5. Calculate Cohen's Kappa for the Pair (Aspect + Sentiment)
pair_dora = merged["aspect_dora"] + "|" + merged["sentiment_dora"]
pair_janice = merged["aspect_janice"] + "|" + merged["sentiment_janice"]
pair_kappa = cohen_kappa_score(pair_dora, pair_janice)

# 6. Print the final results
print("--- Final Kappa Results (Task 61) ---")
print(f"Aspect Kappa:    {aspect_kappa:.4f}")
print(f"Sentiment Kappa: {sentiment_kappa:.4f}")
print(f"Pair Kappa:      {pair_kappa:.4f}")
```
